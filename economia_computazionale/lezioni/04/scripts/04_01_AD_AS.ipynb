```python
import numpy as np
from scipy.optimize import root
```

```python
# Parametri
c = 0.6
b = -1500
k = 0.2
h = 1000
t = 0.2
alpha = 0.5

# Variabili esogene
barA = 1
barK = 30000
barC = 160
barI = 100
barG = 200
barM = 1000
barW = 50
barL = 225
```

```python
# Funzione contenente le equazioni del sistema
def sistema_ad_as(x):
    Y, P = x
    eq1 = Y - ((barC + barI + barG) - (b / h) * (barL - barM / P)) / (1 - c * (1 - t) + b * k / h)
    eq2 = Y - barA * barK**alpha * (barW / (P * barA * barK**alpha * (1 - alpha)))**(-(1 - alpha) / alpha)
    return [eq1, eq2]
```

```python

# Ipotesi iniziale di soluzione
x0 = [1000, 1]

# Soluzione del sistema non lineare
soluzione = root(sistema_ad_as, x0)

# Estrazione delle soluzioni
Y, P = soluzione.x

# Calcolo altre variabili endogene
R = (barL + k * Y - barM / P) / h  # Tasso di interesse
I = barI - b * R  # Investimenti
T = t * Y  # Imposte
C = barC + c * (Y - T)  # Consumo
H = (barW / (P * barA * barK**alpha * (1 - alpha)))**(-1 / alpha)  # Lavoro

# Visualizzazione dei risultati
print(f"Y = {round(Y, 2)}")
print(f"P = {round(P, 2)}")
print(f"R = {round(100 * R, 2)}%")
print(f"I = {round(I, 2)}")
print(f"T = {round(T, 2)}")
print(f"C = {round(C, 2)}")
print(f"H = {round(H, 2)}")
```
